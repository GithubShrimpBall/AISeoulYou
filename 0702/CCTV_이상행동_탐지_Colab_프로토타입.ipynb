## 3. 추론 엔진 생성

아래 셀은 독립 실행 가능한 Python 모듈을 생성합니다. 최신 Ultralytics pose 모델을 먼저 시도하고, 환경에 따라 YOLO11로 자동 대체합니다.


````python
%%writefile cctv_abnormal_prototype.py
"""CCTV abnormal-behavior prototype for Google Colab.

Pipeline:
  YOLO pose + ByteTrack -> per-person temporal features -> interpretable rules
  + optional generic/custom object detector for knife/throwable objects.

Outputs:
  annotated_result.mp4, events.csv, events.json, alerts/*.jpg

This is a high-recall prototype, not a production emergency decision system.
"""
from __future__ import annotations

import json
import math
import os
import re
from collections import defaultdict, deque
from dataclasses import asdict, dataclass
from pathlib import Path
from typing import Any, Deque, Dict, Iterable, List, Optional, Sequence, Tuple

import cv2
import numpy as np
import pandas as pd
import torch
from tqdm.auto import tqdm
from ultralytics import YOLO


@dataclass
class Config:
    # Stable defaults. The loader tries current Ultralytics weights first, then YOLO11.
    pose_model_primary: str = "yolo26n-pose.pt"
    pose_model_fallback: str = "yolo11n-pose.pt"
    object_model_primary: str = "yolo26n.pt"
    object_model_fallback: str = "yolo11n.pt"
    custom_weapon_model: Optional[str] = None

    imgsz: int = 640
    pose_conf: float = 0.30
    object_conf: float = 0.24
    keypoint_conf: float = 0.25
    target_proc_fps: float = 8.0
    tracker: str = "bytetrack.yaml"

    # Prototype thresholds. Calibrate per camera/site.
    fall_threshold: float = 0.72
    throw_threshold: float = 0.72
    loiter_threshold: float = 0.70
    agitation_threshold: float = 0.72
    violence_threshold: float = 0.72
    knife_visible_threshold: float = 0.42
    knife_attack_threshold: float = 0.72

    loiter_min_sec: float = 20.0       # Demo value; production often needs longer.
    loiter_window_sec: float = 35.0
    agitation_window_sec: float = 8.0
    history_sec: float = 70.0
    alert_cooldown_sec: float = 6.0
    active_banner_sec: float = 2.5

    # Optional polygon in normalized coordinates [(x/w, y/h), ...].
    # None means the whole frame, which is intentionally permissive for demos.
    loiter_roi_norm: Optional[List[Tuple[float, float]]] = None

    # Objects used as weak evidence for throwing; custom weapon models may add names.
    throwable_names: Tuple[str, ...] = (
        "bottle", "sports ball", "backpack", "handbag", "suitcase", "cell phone"
    )
    weapon_names: Tuple[str, ...] = (
        "knife", "weapon", "machete", "sword", "gun", "pistol"
    )


EVENT_KO = {
    "FALL": "쓰러짐",
    "THROW_SUSPECTED": "물건 투척 의심",
    "LOITERING_SUSPECTED": "배회·서성임 의심",
    "AGITATION_SUSPECTED": "초조·동요 의심",
    "VIOLENCE_SUSPECTED": "폭력 의심",
    "KNIFE_VISIBLE": "칼/무기 노출 의심",
    "KNIFE_ATTACK_SUSPECTED": "칼부림 위험 의심",
}

# COCO pose indices.
NOSE = 0
L_SHOULDER, R_SHOULDER = 5, 6
L_ELBOW, R_ELBOW = 7, 8
L_WRIST, R_WRIST = 9, 10
L_HIP, R_HIP = 11, 12
L_KNEE, R_KNEE = 13, 14
L_ANKLE, R_ANKLE = 15, 16


def clip01(x: float) -> float:
    return float(np.clip(x, 0.0, 1.0))


def safe_name(text: str) -> str:
    return re.sub(r"[^A-Za-z0-9_.-]+", "_", text)


def model_names(model: YOLO) -> Dict[int, str]:
    names = model.names
    if isinstance(names, dict):
        return {int(k): str(v) for k, v in names.items()}
    return {i: str(v) for i, v in enumerate(names)}


def load_yolo(primary: str, fallback: str) -> YOLO:
    try:
        return YOLO(primary)
    except Exception as exc:
        print(f"[WARN] Could not load {primary}: {exc}\n       Falling back to {fallback}")
        return YOLO(fallback)


def valid_kp(kps: np.ndarray, conf: np.ndarray, idx: int, thr: float) -> bool:
    return (
        kps is not None
        and conf is not None
        and idx < len(kps)
        and idx < len(conf)
        and float(conf[idx]) >= thr
        and np.isfinite(kps[idx]).all()
        and not np.allclose(kps[idx], 0)
    )


def midpoint(
    kps: np.ndarray, conf: np.ndarray, a: int, b: int, thr: float
) -> Optional[np.ndarray]:
    ok_a, ok_b = valid_kp(kps, conf, a, thr), valid_kp(kps, conf, b, thr)
    if ok_a and ok_b:
        return (kps[a] + kps[b]) / 2.0
    if ok_a:
        return kps[a].copy()
    if ok_b:
        return kps[b].copy()
    return None


def torso_angle_from_vertical(kps: np.ndarray, conf: np.ndarray, thr: float) -> float:
    shoulder = midpoint(kps, conf, L_SHOULDER, R_SHOULDER, thr)
    hip = midpoint(kps, conf, L_HIP, R_HIP, thr)
    if shoulder is None or hip is None:
        return float("nan")
    v = shoulder - hip
    norm = float(np.linalg.norm(v))
    if norm < 1e-6:
        return float("nan")
    # 0° = upright, 90° = horizontal. abs handles inverted detections.
    cos_to_vertical = clip01(abs(float(v[1])) / norm)
    return float(np.degrees(np.arccos(cos_to_vertical)))


def angle_3pt(a: np.ndarray, b: np.ndarray, c: np.ndarray) -> float:
    ba, bc = a - b, c - b
    denom = float(np.linalg.norm(ba) * np.linalg.norm(bc))
    if denom < 1e-6:
        return float("nan")
    value = float(np.clip(np.dot(ba, bc) / denom, -1.0, 1.0))
    return float(np.degrees(np.arccos(value)))


def arm_extension(kps: np.ndarray, conf: np.ndarray, thr: float) -> float:
    values: List[float] = []
    for s, e, w in ((L_SHOULDER, L_ELBOW, L_WRIST), (R_SHOULDER, R_ELBOW, R_WRIST)):
        if all(valid_kp(kps, conf, idx, thr) for idx in (s, e, w)):
            ang = angle_3pt(kps[s], kps[e], kps[w])
            if np.isfinite(ang):
                values.append(ang / 180.0)
    return float(max(values, default=0.0))


def hand_near_head(kps: np.ndarray, conf: np.ndarray, bbox_h: float, thr: float) -> float:
    if not valid_kp(kps, conf, NOSE, thr):
        return 0.0
    nose = kps[NOSE]
    radius = max(18.0, 0.25 * bbox_h)
    hits = 0
    valid = 0
    for idx in (L_WRIST, R_WRIST):
        if valid_kp(kps, conf, idx, thr):
            valid += 1
            hits += int(np.linalg.norm(kps[idx] - nose) <= radius)
    return float(hits / max(valid, 1))


def bbox_iou(a: Sequence[float], b: Sequence[float]) -> float:
    ax1, ay1, ax2, ay2 = map(float, a)
    bx1, by1, bx2, by2 = map(float, b)
    ix1, iy1 = max(ax1, bx1), max(ay1, by1)
    ix2, iy2 = min(ax2, bx2), min(ay2, by2)
    iw, ih = max(0.0, ix2 - ix1), max(0.0, iy2 - iy1)
    inter = iw * ih
    union = max(1e-6, (ax2 - ax1) * (ay2 - ay1) + (bx2 - bx1) * (by2 - by1) - inter)
    return float(inter / union)


def point_in_roi_norm(point_xy: Tuple[float, float], roi: Optional[List[Tuple[float, float]]]) -> bool:
    if roi is None:
        return True
    poly = np.asarray(roi, dtype=np.float32)
    return cv2.pointPolygonTest(poly, point_xy, False) >= 0


def direction_changes(points: np.ndarray, min_step: float = 0.004, angle_deg: float = 100.0) -> int:
    if len(points) < 4:
        return 0
    vecs = np.diff(points, axis=0)
    vecs = [v for v in vecs if float(np.linalg.norm(v)) >= min_step]
    changes = 0
    for a, b in zip(vecs[:-1], vecs[1:]):
        denom = float(np.linalg.norm(a) * np.linalg.norm(b))
        if denom < 1e-8:
            continue
        angle = math.degrees(math.acos(float(np.clip(np.dot(a, b) / denom, -1.0, 1.0))))
        changes += int(angle >= angle_deg)
    return changes


def window(history: Deque[Dict[str, Any]], now: float, seconds: float) -> List[Dict[str, Any]]:
    return [s for s in history if s["t"] >= now - seconds]


def reference_before(history: Deque[Dict[str, Any]], target_t: float) -> Optional[Dict[str, Any]]:
    candidates = [s for s in history if s["t"] <= target_t]
    return candidates[-1] if candidates else (history[0] if history else None)


def recent_wrist_speed(
    kps: np.ndarray,
    conf: np.ndarray,
    prev: Optional[Dict[str, Any]],
    t: float,
    bbox_h: float,
    thr: float,
) -> float:
    if prev is None:
        return 0.0
    dt = max(1e-3, t - float(prev["t"]))
    prev_kps, prev_conf = prev["kps"], prev["kp_conf"]
    speeds = []
    for idx in (L_WRIST, R_WRIST):
        if valid_kp(kps, conf, idx, thr) and valid_kp(prev_kps, prev_conf, idx, thr):
            speeds.append(float(np.linalg.norm(kps[idx] - prev_kps[idx])) / max(20.0, bbox_h) / dt)
    return float(max(speeds, default=0.0))


def object_center(obj: Dict[str, Any]) -> np.ndarray:
    x1, y1, x2, y2 = obj["xyxy"]
    return np.asarray([(x1 + x2) / 2.0, (y1 + y2) / 2.0], dtype=np.float32)


def object_near_person_wrist(
    obj: Dict[str, Any], kps: np.ndarray, conf: np.ndarray, bbox_h: float, thr: float
) -> bool:
    center = object_center(obj)
    radius = max(28.0, 0.28 * bbox_h)
    for idx in (L_WRIST, R_WRIST):
        if valid_kp(kps, conf, idx, thr) and np.linalg.norm(center - kps[idx]) <= radius:
            return True
    return False


def parse_objects(result: Any, names: Dict[int, str]) -> List[Dict[str, Any]]:
    if result is None or result.boxes is None or len(result.boxes) == 0:
        return []
    boxes = result.boxes.xyxy.detach().cpu().numpy()
    confs = result.boxes.conf.detach().cpu().numpy()
    classes = result.boxes.cls.detach().cpu().numpy().astype(int)
    out = []
    for box, conf, cls in zip(boxes, confs, classes):
        out.append({"name": names.get(int(cls), str(cls)).lower(), "conf": float(conf), "xyxy": box.tolist()})
    return out


def fall_score(history: Deque[Dict[str, Any]]) -> Tuple[float, Dict[str, float]]:
    if len(history) < 5:
        return 0.0, {}
    now = history[-1]["t"]
    recent = window(history, now, 1.0)
    prior = [s for s in history if now - 2.8 <= s["t"] <= now - 0.65]
    if len(recent) < 3 or len(prior) < 2:
        return 0.0, {}

    horizontal = np.mean([
        (np.isfinite(s["torso_angle"]) and s["torso_angle"] >= 55.0) or s["aspect"] >= 0.95
        for s in recent
    ])
    upright_before = np.mean([
        (np.isfinite(s["torso_angle"]) and s["torso_angle"] <= 35.0) and s["aspect"] <= 0.90
        for s in prior
    ])
    ref = reference_before(history, now - 0.75)
    cur = history[-1]
    drop = 0.0 if ref is None else (cur["center"][1] - ref["center"][1]) / max(20.0, cur["bbox_h"])
    drop_score = clip01((drop - 0.10) / 0.30)
    low_after = 1.0 - clip01(float(np.mean([s["center_speed"] for s in recent])) / 0.40)
    score = 0.42 * horizontal + 0.25 * upright_before + 0.23 * drop_score + 0.10 * low_after
    return float(score), {
        "horizontal": float(horizontal), "upright_before": float(upright_before),
        "drop_bbox_h": float(drop), "low_motion_after": float(low_after)
    }


def throw_score(history: Deque[Dict[str, Any]]) -> Tuple[float, Dict[str, float]]:
    if len(history) < 4:
        return 0.0, {}
    now = history[-1]["t"]
    recent = window(history, now, 1.25)
    if len(recent) < 3:
        return 0.0, {}
    speed = max((s["wrist_speed"] for s in recent), default=0.0)
    speed_score = clip01((speed - 0.75) / 1.75)
    extension = max((s["arm_extension"] for s in recent), default=0.0)
    extension_score = clip01((extension - 0.62) / 0.30)
    held = float(any(s["throwable_near_wrist"] for s in recent[:-1]))
    release = float(held > 0 and not recent[-1]["throwable_near_wrist"])
    upright = float(
        np.isfinite(recent[-1]["torso_angle"]) and recent[-1]["torso_angle"] < 50.0
    )
    score = 0.46 * speed_score + 0.26 * extension_score + 0.18 * max(held, release) + 0.10 * upright
    return float(score), {
        "wrist_speed_bbox_h_s": float(speed), "arm_extension": float(extension),
        "object_held_recently": held, "object_release": release
    }


def loiter_score(
    history: Deque[Dict[str, Any]], cfg: Config
) -> Tuple[float, Dict[str, float]]:
    if len(history) < 4:
        return 0.0, {}
    now = history[-1]["t"]
    recent = window(history, now, cfg.loiter_window_sec)
    duration = recent[-1]["t"] - recent[0]["t"]
    if duration < cfg.loiter_min_sec:
        return 0.0, {"duration_sec": float(duration)}
    pts = np.asarray([s["center_norm"] for s in recent], dtype=np.float32)
    if not point_in_roi_norm(tuple(pts[-1]), cfg.loiter_roi_norm):
        return 0.0, {"duration_sec": float(duration), "outside_roi": 1.0}
    path = float(np.linalg.norm(np.diff(pts, axis=0), axis=1).sum())
    net = float(np.linalg.norm(pts[-1] - pts[0]))
    span = float(max(np.ptp(pts[:, 0]), np.ptp(pts[:, 1])))
    efficiency = net / max(path, 1e-6)
    changes = direction_changes(pts)

    duration_score = clip01(duration / cfg.loiter_min_sec)
    compact_score = 1.0 - clip01(span / 0.28)
    low_eff = 1.0 - clip01(efficiency / 0.60)
    moving_loop = clip01(path / 0.12) * low_eff
    stationary = 1.0 - clip01(path / 0.025)
    pattern = max(moving_loop, 0.72 * stationary)
    change_score = clip01(changes / 5.0)
    score = 0.35 * duration_score + 0.33 * pattern + 0.22 * compact_score + 0.10 * change_score
    return float(score), {
        "duration_sec": float(duration), "path_norm": path, "net_norm": net,
        "path_efficiency": float(efficiency), "span_norm": span,
        "direction_changes": float(changes)
    }


def agitation_score(
    history: Deque[Dict[str, Any]], cfg: Config
) -> Tuple[float, Dict[str, float]]:
    if len(history) < 5:
        return 0.0, {}
    now = history[-1]["t"]
    recent = window(history, now, cfg.agitation_window_sec)
    duration = recent[-1]["t"] - recent[0]["t"]
    if duration < min(5.5, cfg.agitation_window_sec * 0.7):
        return 0.0, {"duration_sec": float(duration)}
    pts = np.asarray([s["center_norm"] for s in recent], dtype=np.float32)
    path = float(np.linalg.norm(np.diff(pts, axis=0), axis=1).sum())
    net = float(np.linalg.norm(pts[-1] - pts[0]))
    efficiency = net / max(path, 1e-6)
    changes = direction_changes(pts)
    hand_head = float(np.mean([s["hand_near_head"] for s in recent]))
    wrist_motion = float(np.mean([min(s["wrist_speed"], 3.0) for s in recent]))
    center_motion = float(np.mean([min(s["center_speed"], 1.5) for s in recent]))

    score = (
        0.29 * clip01(changes / 5.0)
        + 0.24 * clip01(hand_head / 0.35)
        + 0.25 * clip01(wrist_motion / 1.15)
        + 0.12 * clip01(center_motion / 0.55)
        + 0.10 * (1.0 - clip01(efficiency / 0.65))
    )
    return float(score), {
        "duration_sec": float(duration), "direction_changes": float(changes),
        "hand_near_head_ratio": hand_head, "mean_wrist_speed": wrist_motion,
        "path_efficiency": float(efficiency)
    }


def violence_pair_score(a: Dict[str, Any], b: Dict[str, Any]) -> Tuple[float, Dict[str, float]]:
    mean_h = max(30.0, (a["bbox_h"] + b["bbox_h"]) / 2.0)
    distance = float(np.linalg.norm(a["center"] - b["center"])) / mean_h
    proximity = clip01((1.65 - distance) / 1.05)
    wrist_motion = clip01((a["wrist_speed"] + b["wrist_speed"] - 0.7) / 2.2)
    overlap = clip01(bbox_iou(a["bbox"], b["bbox"]) / 0.20)
    score = 0.46 * proximity + 0.37 * wrist_motion + 0.17 * overlap
    return float(score), {
        "distance_bbox_h": distance, "combined_wrist_motion": a["wrist_speed"] + b["wrist_speed"],
        "bbox_iou": bbox_iou(a["bbox"], b["bbox"])
    }


class AlertManager:
    def __init__(self, cfg: Config):
        self.cfg = cfg
        self.hits: Dict[Tuple[str, str], Deque[Tuple[float, float]]] = defaultdict(deque)
        self.last_emit: Dict[Tuple[str, str], float] = defaultdict(lambda: -1e9)
        self.active: List[Dict[str, Any]] = []
        self.events: List[Dict[str, Any]] = []

    def consider(
        self,
        event: str,
        entity_key: str,
        track_ids: Sequence[int],
        score: float,
        threshold: float,
        t: float,
        frame_idx: int,
        details: Dict[str, Any],
        min_hits: int = 2,
        hit_window: float = 1.4,
        cooldown: Optional[float] = None,
    ) -> Optional[Dict[str, Any]]:
        if score < threshold:
            return None
        key = (event, entity_key)
        q = self.hits[key]
        q.append((t, score))
        while q and q[0][0] < t - hit_window:
            q.popleft()
        if len(q) < min_hits:
            return None
        cool = self.cfg.alert_cooldown_sec if cooldown is None else cooldown
        if t - self.last_emit[key] < cool:
            return None
        self.last_emit[key] = t
        item = {
            "event_id": len(self.events) + 1,
            "event": event,
            "event_ko": EVENT_KO.get(event, event),
            "entity_key": entity_key,
            "track_ids": [int(x) for x in track_ids],
            "score": round(float(score), 4),
            "time_sec": round(float(t), 3),
            "frame_idx": int(frame_idx),
            "details": {k: round(float(v), 4) if isinstance(v, (float, np.floating)) else v for k, v in details.items()},
            "snapshot_path": "",
        }
        self.events.append(item)
        self.active.append({"until": t + self.cfg.active_banner_sec, **item})
        return item

    def active_at(self, t: float) -> List[Dict[str, Any]]:
        self.active = [x for x in self.active if x["until"] >= t]
        return self.active


class AbnormalBehaviorEngine:
    def __init__(self, cfg: Optional[Config] = None):
        self.cfg = cfg or Config()
        self.device: Any = 0 if torch.cuda.is_available() else "cpu"
        print(f"[INFO] Device: {self.device}")
        self.pose_model = load_yolo(self.cfg.pose_model_primary, self.cfg.pose_model_fallback)
        if self.cfg.custom_weapon_model:
            self.object_model = YOLO(self.cfg.custom_weapon_model)
            self.using_custom_weapon = True
        else:
            self.object_model = load_yolo(self.cfg.object_model_primary, self.cfg.object_model_fallback)
            self.using_custom_weapon = False
        self.object_names = model_names(self.object_model)
        target_names = set(self.cfg.throwable_names) | set(self.cfg.weapon_names)
        self.target_object_ids = [idx for idx, name in self.object_names.items() if name.lower() in target_names]
        if self.using_custom_weapon or not self.target_object_ids:
            self.target_object_ids = None

        self.histories: Dict[int, Deque[Dict[str, Any]]] = defaultdict(deque)
        self.alerts = AlertManager(self.cfg)

    def _trim_history(self, track_id: int, now: float) -> None:
        q = self.histories[track_id]
        while q and q[0]["t"] < now - self.cfg.history_sec:
            q.popleft()

    def _make_snapshot(
        self,
        track_id: int,
        t: float,
        bbox: np.ndarray,
        kps: np.ndarray,
        kp_conf: np.ndarray,
        objects: List[Dict[str, Any]],
        frame_w: int,
        frame_h: int,
    ) -> Dict[str, Any]:
        x1, y1, x2, y2 = map(float, bbox)
        w, h = max(1.0, x2 - x1), max(1.0, y2 - y1)
        center = np.asarray([(x1 + x2) / 2.0, (y1 + y2) / 2.0], dtype=np.float32)
        history = self.histories[track_id]
        prev = history[-1] if history else None
        dt = max(1e-3, t - prev["t"]) if prev else 1.0
        center_speed = (
            0.0 if prev is None else float(np.linalg.norm(center - prev["center"])) / max(20.0, h) / dt
        )
        throwable_near = False
        knife_near = False
        knife_conf = 0.0
        for obj in objects:
            name = obj["name"].lower()
            near = object_near_person_wrist(obj, kps, kp_conf, h, self.cfg.keypoint_conf)
            if near and name in self.cfg.throwable_names:
                throwable_near = True
            if near and name in self.cfg.weapon_names:
                knife_near = True
                knife_conf = max(knife_conf, float(obj["conf"]))
        return {
            "track_id": int(track_id), "t": float(t), "bbox": [x1, y1, x2, y2],
            "bbox_w": w, "bbox_h": h, "aspect": w / h,
            "center": center, "center_norm": center / np.asarray([frame_w, frame_h], dtype=np.float32),
            "center_speed": center_speed,
            "kps": kps.astype(np.float32), "kp_conf": kp_conf.astype(np.float32),
            "torso_angle": torso_angle_from_vertical(kps, kp_conf, self.cfg.keypoint_conf),
            "arm_extension": arm_extension(kps, kp_conf, self.cfg.keypoint_conf),
            "hand_near_head": hand_near_head(kps, kp_conf, h, self.cfg.keypoint_conf),
            "wrist_speed": recent_wrist_speed(kps, kp_conf, prev, t, h, self.cfg.keypoint_conf),
            "throwable_near_wrist": throwable_near,
            "knife_near_wrist": knife_near,
            "knife_conf": knife_conf,
        }

    def process_video(
        self,
        input_path: str,
        output_path: str = "annotated_result.mp4",
        event_csv: str = "events.csv",
        event_json: str = "events.json",
        alerts_dir: str = "alerts",
    ) -> pd.DataFrame:
        input_path = str(input_path)
        out_path = Path(output_path)
        out_path.parent.mkdir(parents=True, exist_ok=True)
        alerts_path = Path(alerts_dir)
        alerts_path.mkdir(parents=True, exist_ok=True)

        cap = cv2.VideoCapture(input_path)
        if not cap.isOpened():
            raise FileNotFoundError(f"Could not open video: {input_path}")
        src_fps = float(cap.get(cv2.CAP_PROP_FPS) or 30.0)
        width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
        height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
        total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT) or 0)
        frame_step = max(1, int(round(src_fps / self.cfg.target_proc_fps)))
        out_fps = src_fps / frame_step
        writer = cv2.VideoWriter(
            str(out_path), cv2.VideoWriter_fourcc(*"mp4v"), out_fps, (width, height)
        )
        if not writer.isOpened():
            raise RuntimeError(f"Could not create output video: {out_path}")

        print(
            f"[INFO] {width}x{height}, source={src_fps:.2f} FPS, process={out_fps:.2f} FPS, "
            f"frame_step={frame_step}"
        )
        pbar = tqdm(total=math.ceil(total_frames / frame_step) if total_frames else None)
        frame_idx = -1
        processed_idx = -1

        try:
            while True:
                ok, frame = cap.read()
                frame_idx += 1
                if not ok:
                    break
                if frame_idx % frame_step != 0:
                    continue
                processed_idx += 1
                t = frame_idx / src_fps

                pose_res = self.pose_model.track(
                    frame,
                    persist=True,
                    tracker=self.cfg.tracker,
                    conf=self.cfg.pose_conf,
                    imgsz=self.cfg.imgsz,
                    device=self.device,
                    verbose=False,
                )[0]
                obj_res = self.object_model.predict(
                    frame,
                    conf=self.cfg.object_conf,
                    imgsz=self.cfg.imgsz,
                    classes=self.target_object_ids,
                    device=self.device,
                    verbose=False,
                )[0]
                objects = parse_objects(obj_res, self.object_names)

                current: Dict[int, Dict[str, Any]] = {}
                if (
                    pose_res.boxes is not None
                    and pose_res.boxes.id is not None
                    and pose_res.keypoints is not None
                    and pose_res.keypoints.xy is not None
                ):
                    boxes = pose_res.boxes.xyxy.detach().cpu().numpy()
                    ids = pose_res.boxes.id.detach().cpu().numpy().astype(int)
                    kxy = pose_res.keypoints.xy.detach().cpu().numpy()
                    if pose_res.keypoints.conf is not None:
                        kconf = pose_res.keypoints.conf.detach().cpu().numpy()
                    else:
                        kconf = np.ones(kxy.shape[:2], dtype=np.float32)

                    for box, track_id, kps, conf in zip(boxes, ids, kxy, kconf):
                        snap = self._make_snapshot(track_id, t, box, kps, conf, objects, width, height)
                        self.histories[int(track_id)].append(snap)
                        self._trim_history(int(track_id), t)
                        current[int(track_id)] = snap

                emitted: List[Dict[str, Any]] = []
                # Per-track rules.
                for track_id, snap in current.items():
                    hist = self.histories[track_id]
                    fs, fd = fall_score(hist)
                    item = self.alerts.consider(
                        "FALL", str(track_id), [track_id], fs, self.cfg.fall_threshold,
                        t, frame_idx, fd, min_hits=2, hit_window=1.3
                    )
                    if item:
                        emitted.append(item)

                    ts, td = throw_score(hist)
                    item = self.alerts.consider(
                        "THROW_SUSPECTED", str(track_id), [track_id], ts, self.cfg.throw_threshold,
                        t, frame_idx, td, min_hits=1, hit_window=0.8, cooldown=4.0
                    )
                    if item:
                        emitted.append(item)

                    ls, ld = loiter_score(hist, self.cfg)
                    item = self.alerts.consider(
                        "LOITERING_SUSPECTED", str(track_id), [track_id], ls, self.cfg.loiter_threshold,
                        t, frame_idx, ld, min_hits=2, hit_window=3.0, cooldown=15.0
                    )
                    if item:
                        emitted.append(item)

                    ags, agd = agitation_score(hist, self.cfg)
                    item = self.alerts.consider(
                        "AGITATION_SUSPECTED", str(track_id), [track_id], ags, self.cfg.agitation_threshold,
                        t, frame_idx, agd, min_hits=3, hit_window=2.0, cooldown=10.0
                    )
                    if item:
                        emitted.append(item)

                    if snap["knife_near_wrist"]:
                        # Nearest other person, normalized by body height.
                        other_dist = 9.0
                        for other_id, other in current.items():
                            if other_id == track_id:
                                continue
                            d = float(np.linalg.norm(snap["center"] - other["center"])) / max(
                                30.0, (snap["bbox_h"] + other["bbox_h"]) / 2.0
                            )
                            other_dist = min(other_dist, d)
                        proximity = clip01((1.7 - other_dist) / 1.1)
                        motion = clip01((snap["wrist_speed"] - 0.45) / 1.45)
                        knife_conf = clip01(snap["knife_conf"])
                        attack_score = 0.48 * knife_conf + 0.30 * proximity + 0.22 * motion
                        item = self.alerts.consider(
                            "KNIFE_ATTACK_SUSPECTED", str(track_id), [track_id], attack_score,
                            self.cfg.knife_attack_threshold, t, frame_idx,
                            {"knife_conf": knife_conf, "nearest_person_bbox_h": other_dist,
                             "wrist_speed": snap["wrist_speed"]},
                            min_hits=1, hit_window=0.5, cooldown=3.0
                        )
                        if item:
                            emitted.append(item)

                # Pairwise physical conflict rule.
                ids_now = sorted(current)
                for i, a_id in enumerate(ids_now):
                    for b_id in ids_now[i + 1:]:
                        score, detail = violence_pair_score(current[a_id], current[b_id])
                        key = f"{a_id}-{b_id}"
                        item = self.alerts.consider(
                            "VIOLENCE_SUSPECTED", key, [a_id, b_id], score,
                            self.cfg.violence_threshold, t, frame_idx, detail,
                            min_hits=2, hit_window=1.2, cooldown=5.0
                        )
                        if item:
                            emitted.append(item)

                # Any visible weapon/knife gets an immediate attention alert.
                for obj_idx, obj in enumerate(objects):
                    if obj["name"] not in self.cfg.weapon_names:
                        continue
                    score = float(obj["conf"])
                    nearest_id: Optional[int] = None
                    nearest = float("inf")
                    oc = object_center(obj)
                    for track_id, snap in current.items():
                        d = float(np.linalg.norm(oc - snap["center"])) / max(30.0, snap["bbox_h"])
                        if d < nearest:
                            nearest, nearest_id = d, track_id
                    entity = str(nearest_id) if nearest_id is not None else f"obj{obj_idx}"
                    tids = [nearest_id] if nearest_id is not None else []
                    item = self.alerts.consider(
                        "KNIFE_VISIBLE", entity, tids, score, self.cfg.knife_visible_threshold,
                        t, frame_idx, {"object": obj["name"], "object_conf": score,
                                       "nearest_person_bbox_h": nearest},
                        min_hits=1, hit_window=0.5, cooldown=4.0
                    )
                    if item:
                        emitted.append(item)

                annotated = pose_res.plot() if pose_res is not None else frame.copy()
                # Draw selected objects.
                for obj in objects:
                    x1, y1, x2, y2 = map(int, obj["xyxy"])
                    is_weapon = obj["name"] in self.cfg.weapon_names
                    color = (0, 0, 255) if is_weapon else (0, 165, 255)
                    cv2.rectangle(annotated, (x1, y1), (x2, y2), color, 2)
                    cv2.putText(
                        annotated, f"{obj['name']} {obj['conf']:.2f}", (x1, max(20, y1 - 6)),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.55, color, 2, cv2.LINE_AA
                    )

                active = self.alerts.active_at(t)
                banner_y = 28
                for alert in active:
                    text = f"ALERT {alert['event']} score={alert['score']:.2f} ID={alert['entity_key']}"
                    cv2.rectangle(annotated, (4, banner_y - 21), (min(width - 4, 720), banner_y + 5), (0, 0, 0), -1)
                    cv2.putText(annotated, text, (10, banner_y), cv2.FONT_HERSHEY_SIMPLEX,
                                0.66, (0, 0, 255), 2, cv2.LINE_AA)
                    banner_y += 30
                    for tid in alert["track_ids"]:
                        if tid in current:
                            x1, y1, x2, y2 = map(int, current[tid]["bbox"])
                            cv2.rectangle(annotated, (x1, y1), (x2, y2), (0, 0, 255), 4)

                cv2.putText(
                    annotated, f"t={t:.1f}s  processed_fps={out_fps:.1f}",
                    (10, height - 12), cv2.FONT_HERSHEY_SIMPLEX, 0.55, (255, 255, 255), 2, cv2.LINE_AA
                )

                # Save alert snapshots after overlays are drawn.
                for item in emitted:
                    snap_file = alerts_path / (
                        f"{item['event_id']:04d}_{safe_name(item['event'])}_"
                        f"{item['time_sec']:.2f}s_{safe_name(item['entity_key'])}.jpg"
                    )
                    cv2.imwrite(str(snap_file), annotated)
                    item["snapshot_path"] = str(snap_file)
                    # Update the active copy too.
                    for active_item in self.alerts.active:
                        if active_item["event_id"] == item["event_id"]:
                            active_item["snapshot_path"] = str(snap_file)

                writer.write(annotated)
                pbar.update(1)
        finally:
            pbar.close()
            cap.release()
            writer.release()

        events_df = pd.DataFrame(self.alerts.events)
        if not events_df.empty:
            # JSON-like details remain readable in CSV.
            csv_df = events_df.copy()
            csv_df["details"] = csv_df["details"].apply(lambda x: json.dumps(x, ensure_ascii=False))
            csv_df["track_ids"] = csv_df["track_ids"].apply(json.dumps)
        else:
            csv_df = pd.DataFrame(columns=[
                "event_id", "event", "event_ko", "entity_key", "track_ids", "score",
                "time_sec", "frame_idx", "details", "snapshot_path"
            ])
        csv_df.to_csv(event_csv, index=False, encoding="utf-8-sig")
        with open(event_json, "w", encoding="utf-8") as f:
            json.dump({"config": asdict(self.cfg), "events": self.alerts.events}, f, ensure_ascii=False, indent=2)
        print(f"[DONE] Video: {out_path}\n       CSV: {event_csv}\n       JSON: {event_json}\n       Alerts: {alerts_path}")
        return events_df


def run_prototype(
    video_path: str,
    output_path: str = "annotated_result.mp4",
    custom_weapon_model: Optional[str] = None,
    loiter_roi_norm: Optional[List[Tuple[float, float]]] = None,
) -> pd.DataFrame:
    cfg = Config(custom_weapon_model=custom_weapon_model, loiter_roi_norm=loiter_roi_norm)
    engine = AbnormalBehaviorEngine(cfg)
    return engine.process_video(video_path, output_path=output_path)


if __name__ == "__main__":
    import argparse

    parser = argparse.ArgumentParser()
    parser.add_argument("video")
    parser.add_argument("--output", default="annotated_result.mp4")
    parser.add_argument("--weapon-model", default=None)
    args = parser.parse_args()
    run_prototype(args.video, args.output, args.weapon_model)
````

# 선택 사항: 데이터셋으로 Pose-TCN 학습

권장 방식은 시간척도가 다른 행동을 분리하는 것입니다.

- **단기 헤드(3~6초):** `normal, fall, throw, violence, knife_attack`
- **장기 헤드(20~60초):** `normal_long, loitering, agitation`

AI-Hub XML을 매니페스트로 바꾸는 함수와, 영상에서 최대 2명의 자세·전역 이동 특징을 추출해 Temporal CNN을 학습하는 코드를 포함합니다. `throw_candidate`, `agitation_candidate`는 반드시 수동 검수 후 이름을 확정하십시오.


````python
%%writefile pose_tcn_training.py
"""Dataset preparation and Pose-TCN training helpers for the CCTV prototype.

Expected manifest columns:
  video_path,label,start_sec,end_sec,group_id

Recommended two-head training:
  short head: normal, fall, throw, violence, knife_attack (roughly 3-6 s clips)
  long head: normal_long, loitering, agitation (roughly 20-60 s clips)
"""
from __future__ import annotations

import json
import math
import xml.etree.ElementTree as ET
from collections import defaultdict
from pathlib import Path
from typing import Dict, Iterable, List, Optional, Sequence, Tuple

import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from sklearn.metrics import ConfusionMatrixDisplay, classification_report, f1_score
from sklearn.model_selection import GroupShuffleSplit
from torch.utils.data import DataLoader, Dataset
from tqdm.auto import tqdm
from ultralytics import YOLO


AIHUB_LABEL_MAP = {
    "swoon": "fall", "실신": "fall", "전도": "fall",
    "wander": "loitering", "배회": "loitering",
    "assault": "violence", "폭행": "violence",
    "fight": "violence", "싸움": "violence",
    "robbery": "violence", "강도": "violence",
    "datefight": "violence", "데이트폭력": "violence", "추행": "violence",
    # These are weak proxy labels and need manual review before use as 'throw'.
    "vandalism": "throw_candidate", "기물파손": "throw_candidate",
    "dump": "throw_candidate", "투기": "throw_candidate",
    # Drunken behavior can seed agitation candidates, but should not be equated with anxiety.
    "drunken": "agitation_candidate", "주취": "agitation_candidate",
}


def _parse_time(value: Optional[str], default: float = 0.0) -> float:
    if value is None:
        return default
    text = str(value).strip()
    try:
        return float(text)
    except ValueError:
        pass
    parts = text.split(":")
    try:
        nums = [float(x) for x in parts]
        if len(nums) == 3:
            return nums[0] * 3600 + nums[1] * 60 + nums[2]
        if len(nums) == 2:
            return nums[0] * 60 + nums[1]
    except ValueError:
        return default
    return default


def _map_aihub_label(event_name: str) -> Optional[str]:
    lowered = event_name.lower().replace(" ", "")
    for key, value in AIHUB_LABEL_MAP.items():
        if key.lower().replace(" ", "") in lowered:
            return value
    return None


def build_aihub_manifest(
    xml_root: str,
    video_root: str,
    output_csv: str = "aihub_manifest_raw.csv",
) -> pd.DataFrame:
    """Parse the AI-Hub XML fields eventname/starttime/duration into a manifest.

    The released archive can have nested folders, so video files are indexed by stem.
    Rows marked *_candidate require manual relabeling.
    """
    xml_root_p, video_root_p = Path(xml_root), Path(video_root)
    video_index: Dict[str, Path] = {}
    for ext in ("*.mp4", "*.avi", "*.mov", "*.mkv"):
        for path in video_root_p.rglob(ext):
            video_index.setdefault(path.stem, path)

    rows: List[Dict[str, object]] = []
    for xml_path in tqdm(list(xml_root_p.rglob("*.xml")), desc="AI-Hub XML"):
        try:
            root = ET.parse(xml_path).getroot()
        except ET.ParseError:
            continue
        filename = (
            root.findtext(".//filename")
            or root.findtext(".//fileName")
            or xml_path.stem
        )
        video_path = video_index.get(Path(filename).stem)
        if video_path is None:
            continue
        event_nodes = root.findall(".//event") or [root]
        for event in event_nodes:
            event_name = (
                event.findtext("eventname")
                or event.findtext("eventName")
                or root.findtext(".//eventname")
                or root.findtext(".//eventName")
                or ""
            )
            label = _map_aihub_label(event_name)
            if label is None:
                continue
            start = _parse_time(
                event.findtext("starttime")
                or event.findtext("startTime")
                or root.findtext(".//starttime")
            )
            duration = _parse_time(
                event.findtext("duration")
                or root.findtext(".//event/duration"),
                default=5.0,
            )
            end = max(start + max(duration, 1.0), start + 1.0)
            rows.append({
                "video_path": str(video_path),
                "label": label,
                "start_sec": start,
                "end_sec": end,
                # Replace with actor/scene ID if available to prevent data leakage.
                "group_id": video_path.parent.name,
                "source": "AI-Hub",
                "raw_event": event_name,
                "needs_review": label.endswith("_candidate"),
            })
    df = pd.DataFrame(rows)
    df.to_csv(output_csv, index=False, encoding="utf-8-sig")
    print(f"[DONE] {len(df)} rows -> {output_csv}")
    return df


def build_folder_manifest(
    dataset_root: str,
    output_csv: str = "custom_manifest.csv",
) -> pd.DataFrame:
    """Create a simple manifest from dataset_root/<label>/*.mp4.

    Each entire video becomes one clip. Edit start_sec/end_sec to localize the action.
    Use actor/scene/camera identifiers for group_id whenever possible.
    """
    root = Path(dataset_root)
    rows = []
    for label_dir in sorted(p for p in root.iterdir() if p.is_dir()):
        label = label_dir.name
        for ext in ("*.mp4", "*.avi", "*.mov", "*.mkv"):
            for path in label_dir.rglob(ext):
                cap = cv2.VideoCapture(str(path))
                fps = float(cap.get(cv2.CAP_PROP_FPS) or 30.0)
                frames = float(cap.get(cv2.CAP_PROP_FRAME_COUNT) or 0.0)
                cap.release()
                duration = frames / fps if frames > 0 else 5.0
                rows.append({
                    "video_path": str(path), "label": label,
                    "start_sec": 0.0, "end_sec": duration,
                    "group_id": path.stem.split("_")[0],
                    "source": "custom", "raw_event": label, "needs_review": False,
                })
    df = pd.DataFrame(rows)
    df.to_csv(output_csv, index=False, encoding="utf-8-sig")
    print(f"[DONE] {len(df)} rows -> {output_csv}")
    return df


def _load_pose_model(primary: str = "yolo26n-pose.pt", fallback: str = "yolo11n-pose.pt") -> YOLO:
    try:
        return YOLO(primary)
    except Exception:
        return YOLO(fallback)


def _sample_frame_indices(cap: cv2.VideoCapture, start_sec: float, end_sec: float, seq_len: int) -> np.ndarray:
    fps = float(cap.get(cv2.CAP_PROP_FPS) or 30.0)
    total = int(cap.get(cv2.CAP_PROP_FRAME_COUNT) or 0)
    start_f = max(0, int(start_sec * fps))
    end_f = min(total - 1, max(start_f + 1, int(end_sec * fps))) if total > 0 else max(start_f + 1, int(end_sec * fps))
    return np.linspace(start_f, end_f, seq_len).astype(int)


def _frame_pose_features(
    result,
    frame_w: int,
    frame_h: int,
    max_people: int = 2,
) -> np.ndarray:
    """Return [max_people, 55] = 17*(x,y,conf) + bbox(cx,cy,w,h).

    Keypoints are normalized within each person box; bbox features retain global motion.
    Persons are ordered left-to-right after retaining the largest max_people boxes.
    """
    feat_dim = 17 * 3 + 4
    output = np.zeros((max_people, feat_dim), dtype=np.float32)
    if result.boxes is None or result.keypoints is None or result.keypoints.xy is None:
        return output
    boxes = result.boxes.xyxy.detach().cpu().numpy()
    kxy = result.keypoints.xy.detach().cpu().numpy()
    if result.keypoints.conf is not None:
        kconf = result.keypoints.conf.detach().cpu().numpy()
    else:
        kconf = np.ones(kxy.shape[:2], dtype=np.float32)
    if len(boxes) == 0:
        return output
    areas = (boxes[:, 2] - boxes[:, 0]) * (boxes[:, 3] - boxes[:, 1])
    keep = np.argsort(-areas)[:max_people]
    keep = sorted(keep, key=lambda i: float((boxes[i, 0] + boxes[i, 2]) / 2.0))
    for out_i, idx in enumerate(keep):
        x1, y1, x2, y2 = boxes[idx]
        bw, bh = max(1.0, x2 - x1), max(1.0, y2 - y1)
        cx, cy = (x1 + x2) / 2.0, (y1 + y2) / 2.0
        xy = kxy[idx].copy()
        xy[:, 0] = (xy[:, 0] - cx) / bw
        xy[:, 1] = (xy[:, 1] - cy) / bh
        joints = np.concatenate([xy, kconf[idx, :, None]], axis=1).reshape(-1)
        bbox_feat = np.asarray([cx / frame_w, cy / frame_h, bw / frame_w, bh / frame_h], dtype=np.float32)
        output[out_i] = np.concatenate([joints, bbox_feat])
    return output


def extract_pose_sequences(
    manifest_csv: str,
    output_npz: str = "pose_sequences.npz",
    seq_len: int = 48,
    max_people: int = 2,
    imgsz: int = 640,
    conf: float = 0.25,
    include_labels: Optional[Sequence[str]] = None,
) -> str:
    df = pd.read_csv(manifest_csv)
    required = {"video_path", "label", "start_sec", "end_sec", "group_id"}
    missing = required - set(df.columns)
    if missing:
        raise ValueError(f"Manifest missing columns: {sorted(missing)}")
    if include_labels is not None:
        df = df[df["label"].isin(include_labels)].reset_index(drop=True)
    model = _load_pose_model()
    device = 0 if torch.cuda.is_available() else "cpu"
    X, labels, groups, paths = [], [], [], []

    for row in tqdm(df.itertuples(index=False), total=len(df), desc="Pose sequences"):
        cap = cv2.VideoCapture(str(row.video_path))
        if not cap.isOpened():
            continue
        width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH) or 1)
        height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT) or 1)
        indices = _sample_frame_indices(cap, float(row.start_sec), float(row.end_sec), seq_len)
        seq = []
        for frame_idx in indices:
            cap.set(cv2.CAP_PROP_POS_FRAMES, int(frame_idx))
            ok, frame = cap.read()
            if not ok:
                seq.append(np.zeros((max_people, 55), dtype=np.float32))
                continue
            result = model.predict(frame, conf=conf, imgsz=imgsz, device=device, verbose=False)[0]
            seq.append(_frame_pose_features(result, width, height, max_people))
        cap.release()
        X.append(np.stack(seq).reshape(seq_len, -1))
        labels.append(str(row.label))
        groups.append(str(row.group_id))
        paths.append(str(row.video_path))

    if not X:
        raise RuntimeError("No sequences were extracted. Check manifest paths and video codecs.")
    np.savez_compressed(
        output_npz,
        X=np.asarray(X, dtype=np.float32),
        labels=np.asarray(labels), groups=np.asarray(groups), paths=np.asarray(paths),
        seq_len=seq_len, max_people=max_people,
    )
    print(f"[DONE] X={np.asarray(X).shape} -> {output_npz}")
    return output_npz


class PoseSequenceDataset(Dataset):
    def __init__(self, X: np.ndarray, y: np.ndarray):
        self.X = torch.as_tensor(X, dtype=torch.float32)
        self.y = torch.as_tensor(y, dtype=torch.long)

    def __len__(self) -> int:
        return len(self.y)

    def __getitem__(self, idx: int):
        return self.X[idx], self.y[idx]


class ResidualTCNBlock(nn.Module):
    def __init__(self, channels: int, dilation: int, dropout: float = 0.20):
        super().__init__()
        pad = 2 * dilation
        self.net = nn.Sequential(
            nn.Conv1d(channels, channels, kernel_size=5, padding=pad, dilation=dilation),
            nn.BatchNorm1d(channels), nn.ReLU(inplace=True), nn.Dropout(dropout),
            nn.Conv1d(channels, channels, kernel_size=3, padding=dilation, dilation=dilation),
            nn.BatchNorm1d(channels), nn.Dropout(dropout),
        )
        self.act = nn.ReLU(inplace=True)

    def forward(self, x):
        return self.act(x + self.net(x))


class PoseTCN(nn.Module):
    def __init__(self, input_dim: int, num_classes: int, channels: int = 128):
        super().__init__()
        self.stem = nn.Sequential(
            nn.Conv1d(input_dim, channels, kernel_size=5, padding=2),
            nn.BatchNorm1d(channels), nn.ReLU(inplace=True),
        )
        self.blocks = nn.Sequential(
            ResidualTCNBlock(channels, 1),
            ResidualTCNBlock(channels, 2),
            ResidualTCNBlock(channels, 4),
        )
        self.head = nn.Sequential(
            nn.AdaptiveAvgPool1d(1), nn.Flatten(), nn.Dropout(0.25),
            nn.Linear(channels, num_classes),
        )

    def forward(self, x):
        # [B,T,F] -> [B,F,T]
        x = x.transpose(1, 2)
        return self.head(self.blocks(self.stem(x)))


def train_pose_tcn(
    npz_path: str,
    output_model: str = "pose_tcn_best.pt",
    include_labels: Optional[Sequence[str]] = None,
    epochs: int = 30,
    batch_size: int = 32,
    lr: float = 1e-3,
    seed: int = 42,
) -> Dict[str, object]:
    data = np.load(npz_path, allow_pickle=True)
    X = data["X"].astype(np.float32)
    labels = data["labels"].astype(str)
    groups = data["groups"].astype(str)
    if include_labels is not None:
        mask = np.isin(labels, include_labels)
        X, labels, groups = X[mask], labels[mask], groups[mask]
    classes = sorted(np.unique(labels).tolist())
    if len(classes) < 2:
        raise ValueError("At least two classes are required.")
    class_to_idx = {c: i for i, c in enumerate(classes)}
    y = np.asarray([class_to_idx[x] for x in labels], dtype=np.int64)

    splitter = GroupShuffleSplit(n_splits=1, test_size=0.20, random_state=seed)
    train_idx, val_idx = next(splitter.split(X, y, groups=groups))
    mean = X[train_idx].mean(axis=(0, 1), keepdims=True)
    std = X[train_idx].std(axis=(0, 1), keepdims=True) + 1e-5
    X = (X - mean) / std

    train_ds = PoseSequenceDataset(X[train_idx], y[train_idx])
    val_ds = PoseSequenceDataset(X[val_idx], y[val_idx])
    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True, num_workers=2, pin_memory=True)
    val_loader = DataLoader(val_ds, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=True)

    counts = np.bincount(y[train_idx], minlength=len(classes)).astype(np.float32)
    weights = counts.sum() / np.maximum(counts, 1.0)
    weights = weights / weights.mean()
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model = PoseTCN(X.shape[-1], len(classes)).to(device)
    criterion = nn.CrossEntropyLoss(weight=torch.tensor(weights, device=device))
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=1e-4)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=max(1, epochs))

    best_f1 = -1.0
    history = []
    for epoch in range(1, epochs + 1):
        model.train()
        train_loss = 0.0
        for xb, yb in train_loader:
            xb, yb = xb.to(device), yb.to(device)
            optimizer.zero_grad(set_to_none=True)
            loss = criterion(model(xb), yb)
            loss.backward()
            optimizer.step()
            train_loss += float(loss) * len(yb)
        scheduler.step()

        model.eval()
        truth, pred = [], []
        with torch.no_grad():
            for xb, yb in val_loader:
                logits = model(xb.to(device))
                pred.extend(logits.argmax(1).cpu().numpy().tolist())
                truth.extend(yb.numpy().tolist())
        macro_f1 = f1_score(truth, pred, average="macro", zero_division=0)
        row = {"epoch": epoch, "train_loss": train_loss / len(train_ds), "val_macro_f1": macro_f1}
        history.append(row)
        print(f"epoch={epoch:02d} loss={row['train_loss']:.4f} val_macro_f1={macro_f1:.4f}")
        if macro_f1 > best_f1:
            best_f1 = macro_f1
            torch.save({
                "model_state": model.state_dict(), "input_dim": int(X.shape[-1]),
                "classes": classes, "mean": mean.astype(np.float32), "std": std.astype(np.float32),
                "seq_len": int(X.shape[1]), "best_macro_f1": float(best_f1),
            }, output_model)

    checkpoint = torch.load(output_model, map_location=device)
    model.load_state_dict(checkpoint["model_state"])
    model.eval()
    truth, pred = [], []
    with torch.no_grad():
        for xb, yb in val_loader:
            pred.extend(model(xb.to(device)).argmax(1).cpu().numpy().tolist())
            truth.extend(yb.numpy().tolist())
    report = classification_report(truth, pred, target_names=classes, zero_division=0, output_dict=True)
    print(classification_report(truth, pred, target_names=classes, zero_division=0))
    ConfusionMatrixDisplay.from_predictions(truth, pred, display_labels=classes, xticks_rotation=45)
    plt.tight_layout()
    cm_path = str(Path(output_model).with_suffix(".confusion_matrix.png"))
    plt.savefig(cm_path, dpi=160)
    plt.close()
    result = {
        "model": output_model, "classes": classes, "best_macro_f1": best_f1,
        "classification_report": report, "history": history, "confusion_matrix": cm_path,
        "train_size": len(train_idx), "val_size": len(val_idx),
    }
    with open(str(Path(output_model).with_suffix(".metrics.json")), "w", encoding="utf-8") as f:
        json.dump(result, f, ensure_ascii=False, indent=2)
    return result
````

## 9. 현장 보정 체크리스트

1. 카메라별 ROI와 원근 구역을 설정합니다.
2. 임계값은 **사건 단위 Recall, Precision, 시간당 오경보 수**로 보정합니다.
3. 밤·우천·역광·군중·우산·운동·청소 작업을 hard-negative로 넣습니다.
4. 칼은 범용 COCO 모델만으로 운영하지 말고 CCTV 거리·각도에 맞춘 별도 데이터로 재학습합니다.
5. 경보는 5~10초 전후 영상을 묶어 관제요원에게 제공하고, 자동 처벌·신원 판단에는 사용하지 않습니다.
